Layer-1: Compilation Layer
Goal: Accept multiple input types (CSV, Excel, PDF, DOC, Image)
and convert them into structured data for further layers.


```python
import os

folders = [
    "inputs/csv",
    "inputs/excel",
    "inputs/pdf",
    "inputs/doc",
    "inputs/image",
    "outputs/compiled"
]

for f in folders:
    os.makedirs(f, exist_ok=True)

print("Folder structure ready")
import pandas as pd

compiled_data = []

# CSV
for file in os.listdir("inputs/csv"):
    df = pd.read_csv(f"inputs/csv/{file}")
    compiled_data.append(df)

# Excel
for file in os.listdir("inputs/excel"):
    df = pd.read_excel(f"inputs/excel/{file}")
    compiled_data.append(df)

print("CSV & Excel loaded")
compiled_text = []

# PDF
for file in os.listdir("inputs/pdf"):
    compiled_text.append({
        "source": file,
        "type": "pdf",
        "content": "EXTRACTED_TEXT"
    })

# DOC
for file in os.listdir("inputs/doc"):
    compiled_text.append({
        "source": file,
        "type": "doc",
        "content": "EXTRACTED_TEXT"
    })

print("PDF & DOC registered")
image_inputs = []

for file in os.listdir("inputs/image"):
    image_inputs.append({
        "source": file,
        "type": "image",
        "status": "stored_for_ocr"
    })

print("Images registered")
compiled_output = {
    "tabular_data_count": len(compiled_data),
    "text_data_count": len(compiled_text),
    "image_data_count": len(image_inputs),
    "status": "Layer-1 compilation complete"
}
import json

with open("outputs/compiled/layer1_summary.json", "w") as f:
    json.dump(compiled_output, f, indent=2)

print("Layer-1 summary saved")
```
